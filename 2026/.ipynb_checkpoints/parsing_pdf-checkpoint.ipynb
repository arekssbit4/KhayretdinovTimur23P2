```python
def extract_text_from_pdf(pdf_path):
    """Извлечение текста из PDF"""
    doc = fitz.open(pdf_path)
    text = []
    for page in range(len(doc)):
        page_text = doc.load_page(page).get_text("text")
        text.append(page_text)
    doc.close()
    return text
```

```python
def parse_author_improved(text_list):
    """Улучшенный парсинг автора - ищет по всему документу"""
    all_text = ' '.join(text_list)
    
    # 1. Ищем полное имя (Имя Фамилия) - например "Денис Сорокин"
    match = re.search(r'([А-Я][а-я]+\s+[А-Я][а-я]+)\s+@', all_text)
    if match:
        full_name = match.group(1)
        # Ищем ник после него
        nick_match = re.search(rf'{full_name}\s+@([A-Za-z0-9_-]+)', all_text)
        if nick_match:
            return f"{full_name} @{nick_match.group(1)}"
        return full_name
    
    # 2. Ищем паттерн "Имя @ник" (например "Александр @ArtSander")
    match = re.search(r'([А-Яа-яA-Za-z\s]+?)\s+(@[A-Za-z0-9_-]+)', all_text)
    if match:
        return f"{match.group(1).strip()} {match.group(2).strip()}"
    
    # 3. Ищем постранично - строки с ником и датой
    for page_num, page_text in enumerate(text_list):
        lines = page_text.split('\n')
        for i, line in enumerate(lines):
            line_clean = line.strip()
            
            # Паттерн: "Tirarex 1 час назад" - ник и дата в одной строке
            match = re.match(r'^([A-Za-z0-9_-]+)\s+(\d+\s+час)', line_clean)
            if match:
                return match.group(1)
            
            # Паттерн: ник на одной строке, дата на следующей
            if re.match(r'^[A-Za-z0-9_-]{3,30}$', line_clean):
                # Проверяем следующую строку на дату
                if i + 1 < len(lines):
                    next_line = lines[i+1].strip()
                    if re.search(r'\d{1,2}\s+(янв|фев|мар|апр|май|июн|июл|авг|сен|окт|ноя|дек|час)', next_line):
                        return line_clean
                # Проверяем предыдущую строку на дату
                if i > 0:
                    prev_line = lines[i-1].strip()
                    if re.search(r'\d{1,2}\s+(янв|фев|мар|апр|май|июн|июл|авг|сен|окт|ноя|дек|час)', prev_line):
                        return line_clean
    
    # 4. Ищем просто @ник
    match = re.search(r'@([A-Za-z0-9_-]+)', all_text)
    if match:
        return f"@{match.group(1)}"
    
    # 5. Ищем ник без @ (буквы, цифры, дефис, от 3 до 30 символов)
    # Проверяем первые 5 страниц
    for page_num in range(min(5, len(text_list))):
        page_text = text_list[page_num]
        lines = page_text.split('\n')
        for line in lines[:30]:
            line_clean = line.strip()
            if re.match(r'^[A-Za-z0-9_-]{4,30}$', line_clean):
                # Проверяем, что это не техническое слово
                if line_clean.lower() not in ['http', 'https', 'www', 'com', 'ru', 'html', 'pdf', 'docs', 'user', 'guest']:
                    return line_clean
    
    return "Не указан"
```

```python
MONTHS = {
    'янв': 1, 'фев': 2, 'мар': 3, 'апр': 4, 'май': 5, 'июн': 6,
    'июл': 7, 'авг': 8, 'сен': 9, 'окт': 10, 'ноя': 11, 'дек': 12
}

def normalize_date(date_string, current_year=2026):
    """Приводит дату к формату dd.MM.yyyy"""
    if not date_string or date_string == "Не указана":
        return "Не указана"
    
    date_string = date_string.strip()
    
    # 1. dd.MM.yyyy
    match = re.match(r'^(\d{2})\.(\d{2})\.(\d{4})$', date_string)
    if match:
        return date_string
    
    # 2. dd месяц yyyy (7 сен 2022)
    match = re.match(r'^(\d{1,2})\s+(янв|фев|мар|апр|май|июн|июл|авг|сен|окт|ноя|дек)\s+(\d{4})$', date_string)
    if match:
        day = int(match.group(1))
        month = MONTHS[match.group(2)]
        year = int(match.group(3))
        return f"{day:02d}.{month:02d}.{year}"
    
    # 3. dd месяц в HH:MM (23 янв в 16:44) - ЭТО ОСНОВНОЙ ФОРМАТ ДЛЯ ХАБРА
    match = re.match(r'^(\d{1,2})\s+(янв|фев|мар|апр|май|июн|июл|авг|сен|окт|ноя|дек)\s+в\s+\d{2}:\d{2}$', date_string)
    if match:
        day = int(match.group(1))
        month = MONTHS[match.group(2)]
        return f"{day:02d}.{month:02d}.{current_year}"
    
    # 4. dd месяц (17 мар) - без года
    match = re.match(r'^(\d{1,2})\s+(янв|фев|мар|апр|май|июн|июл|авг|сен|окт|ноя|дек)$', date_string)
    if match:
        day = int(match.group(1))
        month = MONTHS[match.group(2)]
        return f"{day:02d}.{month:02d}.{current_year}"
    
    return date_string

def extract_publish_date(text_list, current_year=2026):
    """
    Извлекает ДАТУ ПУБЛИКАЦИИ статьи (не дату создания PDF)
    Приоритет: ищем в первых 100 строках, игнорируем 24.03.2026
    """
    # Берем первые 100 строк документа (здесь должна быть дата публикации)
    first_few_pages = []
    char_count = 0
    for page in text_list:
        first_few_pages.append(page)
        char_count += len(page)
        if char_count > 15000:  # Примерно первые 3-5 страниц
            break
    
    early_text = ' '.join(first_few_pages)
    all_text = ' '.join(text_list)
    
    # Приоритетные паттерны для даты публикации (в порядке важности)
    patterns = [
        # "23 янв в 16:44" - самый частый формат на Хабре
        r'(\d{1,2})\s+(янв|фев|мар|апр|май|июн|июл|авг|сен|окт|ноя|дек)\s+в\s+\d{2}:\d{2}',
        # "4 мар в 14:14" - тоже частый
        r'(\d{1,2})\s+(мар|апр|май|июн|июл|авг|сен|окт|ноя|дек)\s+в\s+\d{2}:\d{2}',
        # "7 сен 2022" - с годом
        r'(\d{1,2})\s+(янв|фев|мар|апр|май|июн|июл|авг|сен|окт|ноя|дек)\s+(\d{4})',
        # "17 мар" - без года
        r'(\d{1,2})\s+(янв|фев|мар|апр|май|июн|июл|авг|сен|окт|ноя|дек)\b',
    ]
    
    for pattern in patterns:
        # Сначала ищем в начале документа
        match = re.search(pattern, early_text)
        if match:
            groups = match.groups()
            if len(groups) == 3 and groups[2].isdigit() and len(groups[2]) == 4:
                # dd месяц yyyy
                day = int(groups[0])
                month = MONTHS[groups[1]]
                year = int(groups[2])
                return f"{day:02d}.{month:02d}.{year}"
            elif len(groups) == 2:
                # dd месяц (без года) или dd месяц в HH:MM
                day = int(groups[0])
                month = MONTHS[groups[1]]
                # Проверяем, есть ли время в этой же строке
                time_match = re.search(r'в\s+\d{2}:\d{2}', match.group(0))
                if time_match:
                    # Это формат "23 янв в 16:44"
                    return f"{day:02d}.{month:02d}.{current_year}"
                else:
                    # Это просто "17 мар"
                    return f"{day:02d}.{month:02d}.{current_year}"
            elif len(groups) == 1:
                # dd месяц yyyy (если год отдельно)
                pass
    
    # Если ничего не нашли
    return "Не указана"
```

```python
def parse_habr_article(text_list, filename, current_year=2026):
    """Парсинг с правильной датой публикации"""
    full_text = ' '.join(text_list).replace('\n', ' ').strip()
    first_page = text_list[0] if text_list else ""
    lines_first = first_page.split('\n')
    
    # 1. Название статьи
    title = ""
    for line in lines_first[:15]:
        if len(line) > 30 and len(line) < 200:
            if any(word in line for word in ['ИИ', 'AI', 'GPT', 'сети', 'базы', 'Искусственный', 
                                              'Дизайн', 'Облака', 'роли', 'отчётов', 'агентов',
                                              'эзотерические', 'гибридные', '5GBE', '2026', '2025']):
                if 'Хабр' not in line and 'Подписаться' not in line and 'Реклама' not in line:
                    title = line.strip()
                    break
    
    if not title:
        title = filename.replace('_ Хабр.pdf', '').replace('PDF\\', '').replace('PDF/', '')
        if len(title) > 100:
            title = title[:97] + "..."
    
    # 2. Автор (используем вашу функцию)
    author = parse_author_improved(text_list)
    
    # 3. ДАТА ПУБЛИКАЦИИ (исправленная)
    publish_date = extract_publish_date(text_list, current_year)
    
    # 4. Рейтинг
    rating = "0"
    rating_patterns = [
        r'(\d+\.?\d*)\s*Общий рейтинг',
        r'(\d+\.?\d*)\s*K\s*просмотров',
        r'(\d+\.?\d*)K',
        r'(\d+)\s*мин\s+(\d+\.?\d*)\s*K',
        r'(\d+\.?\d*)\s*Оценка',
    ]
    for pattern in rating_patterns:
        rating_match = re.search(pattern, first_page)
        if rating_match:
            groups = rating_match.groups()
            rating = groups[-1] if len(groups) > 1 else groups[0]
            if 'K' not in rating and rating.replace('.', '').isdigit():
                rating = f"{rating}K"
            break
    
    # 5. Описание
    description = (full_text[:300] + "...") if len(full_text) > 300 else full_text
    
    return {
        'Название статьи': title,
        'Автор': author,
        'Дата написания': publish_date,
        'Рейтинг': rating,
        'Описание': description,
        'Текст': full_text
    }
```

```python
# Основной цикл парсинга
results = []
errors = []

for pdf_path in all_pdf:
    try:
        print(f"Обработка: {pdf_path.split('/')[-1]}")
        text_content = extract_text_from_pdf(pdf_path)
        article_data = parse_habr_article(text_content, pdf_path.split('/')[-1])
        article_data['Файл'] = pdf_path.split('/')[-1]
        results.append(article_data)
        print(f"Название: {article_data['Название статьи'][:50]}...")
        print(f"Автор: {article_data['Автор']}")
        print(f"Дата: {article_data['Дата написания']}")
        print(f"Рейтинг: {article_data['Рейтинг']}")
        print(f"Текст: {len(article_data['Текст'])} символов\n")
    except Exception as e:
        print(f"Ошибка: {str(e)}\n")
        errors.append({'Файл': pdf_path, 'Ошибка': str(e)})
```

```python
def get_habr_links_from_web(base_url="https://habr.com/ru/articles/", limit=100):
    """
    Получает список ссылок на статьи с главной страницы
    """
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
    }
    
    links = []
    page = 1
    
    print(f"   Сбор ссылок на {limit} статей...")
    
    while len(links) < limit:
        url = base_url if page == 1 else f"{base_url}page{page}/"
        
        try:
            response = requests.get(url, headers=headers)
            soup = BeautifulSoup(response.content, 'html.parser')
            
            page_links = []
            
            # Ищем все h2 с ссылками
            for h2 in soup.find_all('h2'):
                link = h2.find('a')
                if link:
                    href = link.get('href')
                    if href and '/articles/' in href:
                        if not href.startswith('http'):
                            href = 'https://habr.com' + href
                        page_links.append(href)
            
            # Убираем дубликаты
            page_links = list(dict.fromkeys(page_links))
            
            if not page_links:
                print(f"   Достигнут конец. Найдено: {len(links)}")
                break
            
            links.extend(page_links)
            links = list(dict.fromkeys(links))
            
            print(f"   Страница {page}: +{len(page_links)} ссылок. Всего: {len(links)}")
            page += 1
            time.sleep(0.3)
            
        except Exception as e:
            print(f"   Ошибка: {e}")
            break
    
    return links[:limit]
```

```python
def get_article_rating_from_page(soup):
    """
    Извлекает рейтинг со страницы статьи
    """
    # Пробуем разные способы
    rating_selectors = [
        ('span', {'class': 'tm-votes-score'}),
        ('span', {'class': 'votes-score'}),
        ('div', {'class': 'voting-widget__score'}),
        ('span', {'data-test-id': 'vote-score'})
    ]
    
    for tag, attrs in rating_selectors:
        elem = soup.find(tag, attrs)
        if elem:
            text = elem.text.strip()
            # Извлекаем число
            match = re.search(r'(\d+\.?\d*)', text)
            if match:
                return match.group(1)
    
    # Если не нашли - ищем в мета-тегах
    meta_rating = soup.find('meta', {'name': 'rating'})
    if meta_rating and meta_rating.get('content'):
        return meta_rating.get('content')
    
    return "0"
```

```python
def parse_habr_article_from_web(url):
    """
    Парсит отдельную статью с веб-сайта
    """
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
    }
    
    try:
        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.content, 'html.parser')
        
        # 1. Заголовок - через h1
        title = "Не найден"
        title_elem = soup.find('h1')
        if title_elem:
            title = title_elem.text.strip()
        
        # 2. Автор
        author = "Не указан"
        author_elem = soup.find('a', class_='tm-user-info__username')
        if author_elem:
            author = author_elem.text.strip()
        
        # 3. Дата публикации
        publish_date = "Не указана"
        time_elem = soup.find('time')
        if time_elem:
            datetime_attr = time_elem.get('datetime')
            if datetime_attr:
                try:
                    dt = datetime.fromisoformat(datetime_attr.replace('Z', '+00:00'))
                    publish_date = dt.strftime('%d.%m.%Y')
                except:
                    publish_date = time_elem.text.strip()
            else:
                publish_date = time_elem.text.strip()
        
        # 4. Рейтинг (пробуем найти, но может отсутствовать)
        rating = "0"
        rating_selectors = [
            ('span', {'class': 'tm-votes-score'}),
            ('span', {'class': 'votes-score'}),
            ('div', {'class': 'voting-widget__score'}),
            ('span', {'data-test-id': 'vote-score'})
        ]
        
        for tag, attrs in rating_selectors:
            rating_elem = soup.find(tag, attrs)
            if rating_elem:
                rating_text = rating_elem.text.strip()
                match = re.search(r'(\d+\.?\d*)', rating_text)
                if match:
                    rating = match.group(1)
                break
        
        # 5. Полный текст статьи
        text = ""
        content_elem = soup.find('div', id='post-content-body')
        if content_elem:
            text = content_elem.get_text(separator='\n', strip=True)
        
        # 6. Краткое описание
        description = text[:300] + "..." if len(text) > 300 else text
        
        # Извлекаем ID статьи из URL для прогресс-бара
        article_id = url.split('/')[-2] if url.split('/')[-2].isdigit() else '...'
        
        return {
            'Название статьи': title,
            'Автор': author,
            'Дата написания': publish_date,
            'Рейтинг': rating,
            'Описание': description,
            'Текст': text,
            'Источник': 'web'
        }
        
    except Exception as e:
        return None
```

```python
def parse_habr_web_articles(limit=100):
    """
    Главная функция для сбора статей с веб-сайта
    """
    print("="*60)
    print(f"СБОР СТАТЕЙ С ВЕБ-САЙТА HABR (нужно {limit} шт)")
    print("="*60)
    
    # Получаем ссылки
    print("\n1. Получение ссылок...")
    links = get_habr_links_from_web(limit=limit)
    print(f"\n   Найдено ссылок: {len(links)}")
    
    if not links:
        print("   Не удалось найти ссылки!")
        return []
    
    # Парсим статьи
    print(f"\n2. Парсинг статей (это займет ~{len(links) * 0.5 / 60:.0f} минут)...")
    results = []
    
    for i, url in enumerate(links):
        if len(results) >= limit:
            break
        
        # Извлекаем ID статьи
        article_id = url.split('/')[-2] if url.split('/')[-2].isdigit() else '...'
        print(f"   [{i+1}/{len(links)}] Статья {article_id}...", end=" ")
        
        article = parse_habr_article_from_web(url)
        
        if article:
            results.append(article)
            print(f"✓ ({len(results)}/{limit})")
        else:
            print(f"✗")
        
        time.sleep(0.3)
    
    print(f"\n✅ Собрано статей с веба: {len(results)}")
    return results
```

```python
def merge_and_save_data(pdf_dataframe, web_data_list, output_format='both'):
    """
    Объединяет данные из PDF и WEB, сохраняет в Excel и CSV
    
    Параметры:
    - pdf_dataframe: DataFrame с данными из PDF
    - web_data_list: список словарей с данными из WEB
    - output_format: 'csv', 'excel', 'both'
    """
    
    # Создаем DataFrame из веб-данных
    df_web = pd.DataFrame(web_data_list) if web_data_list else pd.DataFrame()
    
    # Добавляем колонку Источник в PDF данные (если её нет)
    if 'Источник' not in pdf_dataframe.columns:
        pdf_dataframe['Источник'] = 'pdf'
    
    # Объединяем DataFrames
    if not df_web.empty:
        df_combined = pd.concat([pdf_dataframe, df_web], ignore_index=True)
    else:
        df_combined = pdf_dataframe
    
    print("\n" + "="*60)
    print("ОБЪЕДИНЕНИЕ ДАННЫХ")
    print("="*60)
    print(f"  Статей из PDF: {len(pdf_dataframe)}")
    print(f"  Статей из WEB: {len(df_web)}")
    print(f"  Всего статей: {len(df_combined)}")
    
    # Статистика по источникам
    if 'Источник' in df_combined.columns:
        print(f"\n  Распределение по источникам:")
        for source, count in df_combined['Источник'].value_counts().items():
            print(f"    - {source}: {count}")
    
    # Сохраняем результаты
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    
    if output_format in ['csv', 'both']:
        csv_filename = f"habr_articles_combined_{timestamp}.csv"
        df_combined.to_csv(csv_filename, index=False, encoding='utf-8-sig')
        print(f"\nCSV сохранен: {csv_filename}")
    
    if output_format in ['excel', 'both']:
        excel_filename = f"habr_articles_combined_{timestamp}.xlsx"
        
        # Сохраняем в Excel с несколькими листами
        with pd.ExcelWriter(excel_filename, engine='openpyxl') as writer:
            # Основной лист со всеми данными
            df_combined.to_excel(writer, sheet_name='Все статьи', index=False)
            
            # Отдельный лист с PDF статьями
            pdf_dataframe.to_excel(writer, sheet_name='PDF статьи', index=False)
            
            # Отдельный лист с WEB статьями
            if not df_web.empty:
                df_web.to_excel(writer, sheet_name='WEB статьи', index=False)
            
            # Статистика
            stats_data = {
                'Показатель': ['Всего статей', 'Из PDF', 'Из WEB', 'Средний размер текста (символов)', 'Уникальные авторы'],
                'Значение': [
                    len(df_combined),
                    len(pdf_dataframe),
                    len(df_web),
                    round(df_combined['Текст'].str.len().mean(), 0),
                    df_combined['Автор'].nunique()
                ]
            }
            df_stats = pd.DataFrame(stats_data)
            df_stats.to_excel(writer, sheet_name='Статистика', index=False)
        
        print(f"Excel сохранен: {excel_filename}")
    
    return df_combined
```
